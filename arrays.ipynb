## 1) Dimension of array

```python
arr1 = np.array(123)
arr2 = np.array([[1,2],[1,2]])
arr3 = np.array([[[1,2],[1,2]],[[1,2],[1,2]]])
arr4 = np.array([[[[1,2],[1,2]],[[1,2],[1,2]],[[1,2],[1,2]]]])
arr5 = np.array([[[[[[[[1,2,3]]]]]]]])

print("Dimension: ",arr1.ndim)
print("Dimension: ",arr2.ndim)
print("Dimension: ",arr3.ndim)
print("Dimension: ",arr4.ndim)
print("Dimension: ",arr5.ndim)
```

```python
arr5 = np.array([1,2,3], ndmin=6)
print("Dimensions: ",arr5.ndim)
```
